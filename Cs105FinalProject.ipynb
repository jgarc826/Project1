```python
import pandas as pd
from google.colab import files

# Suppose the file is named 'data.csv'
df = pd.read_csv('ClimateData2022_2025.csv')

# Check the first few rows
df.head()
```

**Data Cleaning**

Step 1: Change the Variable Names to more readable names

Step 2: Convert the temperatures to Farenheight as they are in Celcuis.

Step 3: We found that the Riverside dataset is missing values for wind direction, peak wind gust speed, and sunshine duration. Because these variables use 0 as a meaningful value, we cannot safely replace the missing values with zero. Instead of introducing inaccurate or misleading data, we will remove these variables entirely from the dataset and proceed only with the features that have complete and reliable observations. However, for snow, we may fill in with the value 0 since it does not snow in Riverside.(Need to ask for opinion from TA or Professor; for now it is dropped.)

Step 4: We will be dropping any observations that only have the "Date' but are missing every other single column value.

Note: It may be worth finding a dataset that has the values of time of sun/wind direction/ peak wind gust speed

```python

#Data Cleaning: Step 1
df.columns = ['Date','avgTemp', 'minTemp', 'maxTemp', 'precipitation', 'snow','windDirection', 'windSpd', 'wpgt', 'AirPressure', 'tsun']
df.head()
```

```python
#Data Cleaning: Step 2
df['avgTemp'] = df['avgTemp'] * 9/5 + 32
df['minTemp'] = df['minTemp'] * 9/5 + 32
df['maxTemp'] = df['maxTemp'] * 9/5 + 32
```

```python
#Data Cleaning: Step 3
df = df.drop(columns=['snow','windDirection', 'wpgt', 'tsun'])

```

```python
#Quick view at our dataset to ensure everything looks right
df.head()
```

```python
df.isna().sum()
```

**Note**: Since each column has some missing variables, we must discuss what to do with each variable and how to fill in its missing value.



```python
#Data Cleaning: Step 4
#While inspecting the dataset, I found that some observations near the bottom contained only the date, with all other weather variables missing.
#This happened because the dataset is dynamic and updates continuously, so future dates may be listed before their measurements are recorded.
#To remove these incomplete observations, I dropped any rows where all non-date columns were missing:
df = df.dropna(subset=df.columns.drop('Date'), how='all')
```
